```python
# @title
from fractions import Fraction
DIGITOS = "0123456789ABCDEF"

# Pedir y validar una base entre 2 y 16

def leer_base(mensaje):
    while True:
        entrada = input(mensaje)

        try:
            base = int(entrada)

            if 2 <= base <= 16:
                return base

            print("Error: la base debe estar entre 2 y 16.")

        except ValueError:
            print("Error: debes ingresar un numero entero.")

# Convertir un caracter a su valor numerico, por ejemplo de letras a números
# Por ejemplo: de A a 10, F a 15, etc.

def valor_digito(caracter):
    caracter = caracter.upper()

    for i in range(len(DIGITOS)):
        if DIGITOS[i] == caracter:
            return i

    return -1

# Validar que el numero pertenezca a la base indicada

def validar_numero(numero, base):
    if numero == "":
        return False

    # Permitir signo negativo al inicio
    if numero[0] == "-":
        numero = numero[1:]

    if numero == "":
        return False

    # Solamente puede existir un punto decimal
    puntos = 0

    for caracter in numero:

        if caracter == ".":
            puntos += 1

            if puntos > 1:
                return False

        else:
            valor = valor_digito(caracter)

            # Caracter desconocido
            if valor == -1:
                return False

            # El digito no pertenece a la base
            if valor >= base:
                return False

    # Evitar entradas "."
    if numero == ".":
        return False

    return True

#  Convertir de la base de origen a valor numerico
#  También calcula manualmente usando valor posicional.

def convertir_a_valor(numero, base):
    negativo = False

    if numero[0] == "-":
        negativo = True
        numero = numero[1:]

    # Separar parte entera y fraccionaria
    if "." in numero:
        parte_entera, parte_fraccionaria = numero.split(".")
    else:
        parte_entera = numero
        parte_fraccionaria = ""


    # Parte entera. Por ejemplo: 1011 base 2
    # (((1)*2 + 0)*2 + 1)*2 + 1

    valor_entero = 0

    for caracter in parte_entera:
        digito = valor_digito(caracter)
        valor_entero = valor_entero * base + digito

    # Parte fracionaria. Por ejemplo: .101 base 2
    # 1/2 + 0/4 + 1/8

    valor_fraccionario = Fraction(0, 1)
    potencia = base

    for caracter in parte_fraccionaria:
        digito = valor_digito(caracter)

        valor_fraccionario += Fraction(digito, potencia)

        potencia *= base

    valor_total = Fraction(valor_entero, 1) + valor_fraccionario

    if negativo:
        valor_total *= -1

    return valor_total

# Convertir la parte entera mediante divisiones sucesivas

def convertir_entero(valor, base_destino):
    if valor == 0:
        return "0"

    resultado = ""

    while valor > 0:

        residuo = valor % base_destino

        resultado = DIGITOS[residuo] + resultado

        valor = valor // base_destino

    return resultado


#  Convertir la parte fraccionaria mediante multiplicaciones sucesivas

def convertir_fraccion(fraccion, base_destino, precision=12):
    resultado = ""

    contador = 0

    while fraccion != 0 and contador < precision:

        # Multiplicacion sucesiva
        fraccion *= base_destino

        # Obtener parte entera
        digito = fraccion.numerator // fraccion.denominator

        resultado += DIGITOS[digito]

        # Eliminar parte entera
        fraccion -= digito

        contador += 1

    return resultado


# Convertir el numero completo a la base destino

def convertir_a_base_destino(valor, base_destino, precision=12):
    negativo = valor < 0

    if negativo:
        valor = -valor

    # Obtener parte entera
    parte_entera = valor.numerator // valor.denominator

    # Obtener parte fraccionaria
    parte_fraccionaria = valor - parte_entera

    # Conversion mediante divisiones sucesivas
    entero_convertido = convertir_entero(
        parte_entera,
        base_destino
    )

    # Si no tiene fraccion
    if parte_fraccionaria == 0:

        if negativo:
            return "-" + entero_convertido

        return entero_convertido

    # Conversion mediante multiplicaciones sucesivas
    fraccion_convertida = convertir_fraccion(
        parte_fraccionaria,
        base_destino,
        precision
    )

    resultado = entero_convertido + "." + fraccion_convertida

    if negativo:
        resultado = "-" + resultado

    return resultado


# Programa principal (interfaz)
def main():

    print("CONVERSOR DE BASES NUMERICAS")

    print()
    print("Bases permitidas: 2 a 16")
    print("Digitos para bases mayores a 10: A-F")
    print("Para numeros fraccionarios utiliza punto decimal.")
    print()

    # Pedir base de origen
    base_origen = leer_base(
        "Ingresa la base de origen (2-16): "
    )

    # Pedir numero hasta que sea valido
    while True:

        numero = input(
            f"Ingresa el numero en base {base_origen}: "
        ).strip()

        if validar_numero(numero, base_origen):
            break

        print()
        print("ERROR: El numero ingresado no es valido")
        print(f"para la base {base_origen}.")
        print("Intenta nuevamente.")
        print()

    # Pedir base destino
    base_destino = leer_base(
        "Ingresa la base de destino (2-16): "
    )

    # Convertir manualmente el numero de origen
    valor = convertir_a_valor(
        numero,
        base_origen
    )

    # Convertir hacia la base destino
    resultado = convertir_a_base_destino(
        valor,
        base_destino
    )

    print()

    print("RESULTADO")
    print(
        f"{numero.upper()} en base {base_origen}"
    )
    print("es igual a")
    print(
        f"{resultado} en base {base_destino}"
    )

# Ejecutar programa
if __name__ == "__main__":
    main()
```

```python

def sumar_un_bit(bit_a, bit_b, acarreo_in):
    # Calcula la suma aritmética de los bits y el acarreo previo
    suma_total = bit_a + bit_b + acarreo_in

    suma = suma_total % 2

    if suma_total >= 2:
        acarreo_out = 1
    else:
        acarreo_out = 0
    # Si la suma es 2 o 3 (ej. 1+1+0 o 1+1+1), se genera un acarreo de salida (Carry Out)
    return suma, acarreo_out


def convertir_a_5_bits(numero):
    # Inicializa un arreglo para los 5 bits de magnitud
    bits = [0, 0, 0, 0, 0]

    # Empezamos desde el bit menos significativo
    # y avanzamos hacia la izquierda.

    for i in range(4, -1, -1):

        bits[i] = numero % 2

        numero = numero // 2

    return bits


def decimal_a_binario_6_bits(numero):

    if numero >= 0:

        magnitud = convertir_a_5_bits(numero)

        # Agregamos 0 como bit de signo
        bits = [0] + magnitud

        return bits


    if numero == -32:

        return [1, 0, 0, 0, 0, 0]


    magnitud = -numero

    # Convertir magnitud a 5 bits
    bits_magnitud = convertir_a_5_bits(magnitud)

    # Inversión de bits (Complemento a 1)
    bits_invertidos = []

    for bit in bits_magnitud:

        if bit == 0:
            bits_invertidos.append(1)

        else:
            bits_invertidos.append(0)

    # Se suma 1 al LSB para obtener el Complemento a2 final
    acarreo = 1

    for i in range(4, -1, -1):

        suma, acarreo = sumar_un_bit(
            bits_invertidos[i],
            0,
            acarreo
        )

        bits_invertidos[i] = suma

  # Se agrega el bit de signo negativo (1) al inicio
    bits = [1] + bits_invertidos

    return bits


def sumar_6_bits(bits_a, bits_b):

    resultado = [0, 0, 0, 0, 0, 0]

    acarreo = 0

  # Suma los primeros 5 bits (magnitud), propagando el acarreo de derecha a izquierda
    for i in range(5, 0, -1):

        suma, acarreo = sumar_un_bit(
            bits_a[i],
            bits_b[i],
            acarreo
        )

        resultado[i] = suma

    # Guarda el acarreo que entra al bit más significativo (bit de signo)
    acarreo_entrada_signo = acarreo
    # Suma final correspondiente al bit de signo
    suma_signo, acarreo_salida_signo = sumar_un_bit(
        bits_a[0],
        bits_b[0],
        acarreo_entrada_signo
    )

    resultado[0] = suma_signo

    # Detección de Desbordamiento (Overflow):
    # Ocurre un desbordamiento si el acarreo que ENTRA al bit de signo es diferente
    if acarreo_entrada_signo != acarreo_salida_signo:
        overflow = True

    else:
        overflow = False


    return (
        resultado,
        overflow,
        acarreo_entrada_signo,
        acarreo_salida_signo
    )



def bits_a_texto(bits):

    resultado = ""

    for bit in bits:

        resultado = resultado + str(bit)

    return resultado

# Función para la validación de la entrada, asegurando que se respete el rango de 6 bits con signo
def leer_numero(mensaje):

    while True:

        entrada = input(mensaje)

        try:

            numero = int(entrada)
            # El rango válido para complemento a 2 con 6 bits es de -2^(n-1) hasta 2^(n-1)-1 (-32 a 31)
            if -32 <= numero <= 31:

                return numero

            else:

                print(
                    "ERROR: El numero debe estar "
                    "entre -32 y 31."
                )

        except ValueError:

            print(
                "ERROR: Debes ingresar un numero entero."
            )

def main():

    print("      SUMADOR BINARIO CON SIGNO DE 6 BITS")
    print()
    print("Rango permitido: -32 a 31")
    print()

  # Petición de los dos operandos decimales
    numero_a = leer_numero(
        "Ingresa el primer numero decimal: "
    )

    numero_b = leer_numero(
        "Ingresa el segundo numero decimal: "
    )


  # Transformación a los arreglos de bits correspondientes
    bits_a = decimal_a_binario_6_bits(numero_a)

    bits_b = decimal_a_binario_6_bits(numero_b)


 #Ejecución de la suma binaria
    resultado, overflow, carry_in, carry_out = sumar_6_bits(
        bits_a,
        bits_b
    )


    print()
    print("            REPRESENTACION BINARIA")

    print(
        f"A = {numero_a:3}: {bits_a_texto(bits_a)}"
    )

    print(
        f"B = {numero_b:3}: {bits_a_texto(bits_b)}"
    )

    print(
        f"Resultado: {bits_a_texto(resultado)}"
    )

    print()
    print("Acarreo que entra al bit de signo :",
          carry_in)

    print("Acarreo que sale del bit de signo :",
          carry_out)

    print()

    if overflow:

        print("OVERFLOW: SI")

        print(
            "El resultado excede el rango "
            "representable de 6 bits."
        )

    else:

        print("OVERFLOW: NO")


if __name__ == "__main__":

    main()
```
